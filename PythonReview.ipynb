## Part 3 – Calculations

```python
num1=int(input("Enter the first number: "))
num2=int(input("Enter the second number: "))
num3=int(input("Enter the third number: "))

sum = num1 + num2 + num3
print("The sum of the numbers is: ",sum)

avg = sum/3
print("The average of the numbers is: ",avg)

if(num1>=num2) and (num2>=num3):
    max = num1
elif(num2>=num1) and (num2>=num3):
    max = num2
else:
    max = num3
print("The largest number is: ",max)

if(num1<=num2) and (num2<=num3):
    min = num1
elif(num2<=num1) and (num2<=num3):
    min = num2
else:
    min = num3
print("The smallest number is: ",min)
```

## Part 4 – Decision Making

```python
score = int(input("Enter grade: "))
if score>= 90:
    print("You have recieved an A!")
elif score>= 80:
    print("You have recieved a B!")
elif score>= 70:
    print("You have recieved a C.")
elif score>= 60:
    print("You have recieved a D...")
else: 
    print("You have failed the course.")
```

## Part 6 – Functions

```python
def calculate_average(nums):
    sum2 = sum(nums)
    average = sum2/len(nums)
    
    return average

n = calculate_average([20, 10, 8, 19, 13, 26])
print("The total average is: ",n)
```
